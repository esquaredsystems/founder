## 🔄 Loops
When repeating something multiple times, Python gives you more than one ways.

```python
# We use for loops when the number of repetitions is known
leap_years = 0
birth_year = 2007
current_year = 2025
for year in range(birth_year, current_year):
  if year % 4 == 0:
    leap_years = leap_years + 1

print(f"{leap_years} leap years have occurred since your birth to date")
```

Let's say your dream car is worth 4.5M (PKR) today and appreciates 3% every year.
You want to estimate how long will it take for you to buy one if you save 50K every month.

```python
# Input name, age, dream car, its price, your monthly income and percentage that goes into savings
name = input("Enter your name: ")
age = int(input("Your age: "))
dream_car = input("Which car do you dream about? ")
car_price = int(input("What does it cost? "))
monthly_income = int(input("Your monthly income: "))
percentage_of_savings = float(input("Percentage that goes into savings: "))

# Write a program to determine when you'll be able to purchase your dream car
total_savings = 0
months = 0
while total_savings < car_price:
  # Keep saving money
  saving = monthly_income * percentage_of_savings / 100
  total_savings = total_savings + saving
  months = months + 1

print(f"Hey, {name}, it will take you {months} months to be able to purchase {dream_car}. By that time, you'll be {age + (months // 12)} years old.")
```

```python
# A program which inputs student names until exited
students = []
ages = []
dream_cars = []

while True:
  name = input("Enter the student name: ")
  students.append(name)
  age = int(input("Enter the age: "))
  ages.append(age)
  dream_car = input("Enter the dream car (Make and model): ")
  dream_cars.append(dream_car)
  carry_on = input("Do you want to enter another record? (Y/N)")
  if carry_on == 'N':
    break

print(students)
print(ages)
print(dream_cars)
```

```python
students = []

while True:
  record = {}
  name = input("Enter the student name: ")
  record['name'] = name
  age = int(input("Enter the age: "))
  record['age'] = age
  dream_car = input("Enter the dream car (Make and model): ")
  record['dream_car'] = dream_car
  students.append(record)
  carry_on = input("Do you want to enter another record? (Y/N)")
  if carry_on == 'N':
    break

print(students)
```

```python
for i in range(0, len(students)):
  student_dict = students[i]
  print(f"Dream car of {student_dict['name']} is {student_dict['dream_car']}")
```

```python
# Write a program, which inputs user's current salary and calculates annual income tax on it

salary = int(input("Enter your monthly salary in PKR: "))
tax = 0
annual_salary = salary * 12

# According to FBR, no tax is deducted on an annual salary <= 600000
if annual_salary < 600000:
  tax = 0
else:
  tax = annual_salary * 0.35

print(f"Your current annual salary is {annual_salary}. The income tax payable is {tax}.")
```
